```python
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.primitives import padding
from cryptography.hazmat.backends import default_backend
import os
import base64
```

```python
# === Fungsi Enkripsi ===
def encrypt_aes(plaintext, key):
    # ubah plaintext ke bytes
    data = plaintext.encode()

    # buat IV (Initialization Vector) acak 16 byte
    iv = os.urandom(16)

    # buat objek cipher AES-CBC
    cipher = Cipher(algorithms.AES(key), modes.CBC(iv), backend=default_backend())
    encryptor = cipher.encryptor()

    # tambahkan padding agar panjang data kelipatan 16
    padder = padding.PKCS7(128).padder()
    padded_data = padder.update(data) + padder.finalize()

    # enkripsi data
    ciphertext = encryptor.update(padded_data) + encryptor.finalize()

    # gabungkan IV + ciphertext, lalu ubah ke base64 agar mudah disimpan
    return base64.b64encode(iv + ciphertext).decode()
```

```python
# === Fungsi Dekripsi ===
def decrypt_aes(ciphertext_b64, key):
    # decode dari base64 ke bytes
    ciphertext = base64.b64decode(ciphertext_b64)

    # pisahkan IV dan data terenkripsi
    iv = ciphertext[:16]
    actual_ciphertext = ciphertext[16:]

    # buat objek cipher AES-CBC
    cipher = Cipher(algorithms.AES(key), modes.CBC(iv), backend=default_backend())
    decryptor = cipher.decryptor()

    # dekripsi
    padded_plain = decryptor.update(actual_ciphertext) + decryptor.finalize()

    # hapus padding
    unpadder = padding.PKCS7(128).unpadder()
    plaintext = unpadder.update(padded_plain) + unpadder.finalize()
    return plaintext.decode()
```

```python
# === Main Program ===
if __name__ == "__main__":
    print("=== Aplikasi Kriptografi Sederhana (AES-256 CBC) ===")
    teks = input("Masukkan teks yang ingin dienkripsi: ")

    # buat key acak 32 byte (256-bit)
    key = os.urandom(32)
    print("\nKunci (base64):", base64.b64encode(key).decode())

    # proses enkripsi
    encrypted = encrypt_aes(teks, key)
    print("\nHasil Enkripsi (base64):", encrypted)

    # proses dekripsi
    decrypted = decrypt_aes(encrypted, key)
    print("\nHasil Dekripsi:", decrypted)
```
